```python
SKIP_AFTER = "2026-01-31_00-00-00-000000"
root = '/home/jing/Desktop/realistic-al/experiments/activelearning/cifar10_imb'
root_dir = Path(root)
for regime_dir in root_dir.iterdir():
    for query_method_dir in regime_dir.iterdir():
        if '0.5' not in query_method_dir.name: continue
        count = 0
        for run_dir in query_method_dir.iterdir():
            run_time = datetime.datetime.strptime(run_dir.name, "%Y-%m-%d_%H-%M-%S-%f")
            skip_time = datetime.datetime.strptime(SKIP_AFTER, "%Y-%m-%d_%H-%M-%S-%f")
            if run_time < skip_time:
                continue
            test_metrics_path = run_dir / 'test_metrics.csv'
            if test_metrics_path.exists():
                count += 1
        if count != 3:
            print("experiment does not have metrics for three runs: ", query_method_dir.name, f"count = {count}")
        else:
            print("Exactly three runs:", regime_dir.name + '/' + query_method_dir.name)
```

```python
for regime in regimes:
    regime_dir = root_dir / f'active-cifar10_{regime}'
    if not regime_dir.exists():
        print(f"{regime_dir} not found")
        continue
    
    # List experiment folders
    for exp_dir in regime_dir.iterdir():
        if not exp_dir.is_dir():
            print(f"{exp_dir} not found")
            continue
            
        # Extract query method
        # Naming convention: basic_model-resnet_drop-0_aug-cifar_randaugmentMC_acq-{query}_ep-200
        try:
            query_start = exp_dir.name.find('acq-') + 4
            if query_start == 3: # 'acq-' not found
                 continue
            query_end = exp_dir.name.find('_', query_start)
            if query_end == -1:
                 query = exp_dir.name[query_start:]
            else:
                 query = exp_dir.name[query_start:query_end]
        except Exception:
            print(f"error in {exp_dir}: query not found")
            continue
            
        # Loop through run folders (timestamps)
        for run_dir in exp_dir.iterdir():
            if not run_dir.is_dir() or not run_dir.name[0].isdigit(): # Simple check for timestamp folder assuming it starts with digit
                print(f"{run_dir} skipped")
                continue
                
            test_metrics_path = run_dir / 'test_metrics.csv'
            failed = False
            
            if not test_metrics_path.exists():
                failed = True
            else:
                try:
                    df = pd.read_csv(test_metrics_path)
                    if len(df) < 10:
                        failed = True
                except Exception:
                    failed = True
            
            if failed:
                config_path = run_dir / '.hydra' / 'config.yaml'
                seed = 'N/A'
                if config_path.exists():
                    try:
                        with open(config_path, 'r') as f:
                            cfg = yaml.safe_load(f)
                            seed = cfg.get('trainer', {}).get('seed', 'N/A')
                    except Exception:
                        pass
                
                failed_runs.append({
                    'Regime': regime,
                    'Query': query,
                    'Seed': seed,
                    'Path': str(run_dir)
                })

print(f'Found {len(failed_runs)} failed runs.')
df_failed = pd.DataFrame(failed_runs)
if not df_failed.empty:
    print(df_failed[['Regime', 'Query', 'Seed']])
    # display(df_failed)
```

```python
import os
from pathlib import Path
from datetime import datetime

# Set the path and cutoff date
root_dir = Path('/home/jing/Desktop/realistic-al/experiments/activelearning/cifar10')
cutoff_date = datetime(2026, 1, 27)

print(f"Listing directories created after {cutoff_date.strftime('%Y-%m-%d')}...\n")

regimes = ['high', 'med', 'low']
total_matched_count = 0
matched_experiments = []

for regime in regimes:
    regime_path = root_dir / f'active-cifar10_{regime}'
    if not regime_path.exists():
        continue
        
    # Loop through experiment configurations
    for exp_dir in regime_path.iterdir():
        if not exp_dir.is_dir() or len(list(exp_dir.iterdir())) == 0:
            continue
        exp_match_count = 0
        exp_matched = []
            
        # Loop through individual runs (timestamp folders)
        for run_dir in exp_dir.iterdir():
            if not run_dir.is_dir():
                continue
                
            try:
                # Try to parse the directory name as a timestamp
                # Format: YYYY-MM-DD_HH-MM-SS-ffffff (e.g. 2026-01-29_03-37-37-812348)
                run_date = datetime.strptime(run_dir.name, "%Y-%m-%d_%H-%M-%S-%f")
                
                if run_date > cutoff_date and 'test_metrics.csv' in [child.name for child in run_dir.iterdir()]:
                    # print(f"[{run_date}] {run_dir}")
                    total_matched_count += 1
                    exp_match_count += 1
                    exp_matched.append(run_date)
                    
            except ValueError:
                # directory name doesn't match the timestamp format, skip it
                continue
        
        if exp_match_count > 0:
            print(f"{exp_dir.parent.name}/{exp_dir.name}")
            matched_experiments.append(exp_dir)
            for child in exp_matched:
                print(f"    {child}")

for exp in matched_experiments:
    print(f"{exp.parent.name}/{exp.name}")

print(f"\nTotal directories found: {total_matched_count}")
```
